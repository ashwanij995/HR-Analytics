```python
import sqlite3
import pandas as pd
df = pd.read_csv("C:\\Users\\ashwa\\HR Analytics Project\\data\\HRanalytics.csv")
conn = sqlite3.connect("hr_analytics.db")

df.to_sql(
    "employees",
    conn,
    if_exists="replace",
    index=False
)

print("Data successfully loaded into SQL database!")
```

```python
query = """SELECT Department, COUNT(*) AS employee_count
FROM employees
GROUP BY Department
ORDER BY employee_count DESC;"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """SELECT Attrition, COUNT(*) AS employee_count
FROM employees
GROUP BY Attrition;"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """SELECT Attrition, ROUND(AVG(MonthlyIncome), 2) AS average_income
FROM employees
GROUP BY Attrition;"""

result = pd.read_sql_query(query, conn)

result
```

```python
query = """SELECT
    Department,
    COUNT(*) AS total_employees,
    SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END) AS employees_left,
    ROUND(
        100.0 * SUM(CASE WHEN Attrition = 'Yes' THEN 1 ELSE 0 END)
        / COUNT(*), 2
    ) AS attrition_rate
FROM employees
GROUP BY Department
ORDER BY attrition_rate DESC;"""

result = pd.read_sql_query(query, conn)

result
```
